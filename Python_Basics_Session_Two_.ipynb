
**3. Lists**
Lists are ordered collections of items. They can hold items of different data types (though it's common for them to hold items of the same type).
Lists are mutable, meaning you can change their content after they are created.
Lists are defined using square brackets `[]`.

**Key List Operations:**
*   Creating an empty list: `my_list = []`
*   Adding items: `my_list.append(item)`
*   Accessing items by index (0-based): `my_list[0]` (first item), `my_list[-1]` (last item)
*   Slicing (getting a sub-list): `my_list[start:end]` (gets items from `start` up to, but not including, `end`)
*   Looping through a list: `for item in my_list:`

**Example:**



```python
# Create an empty list
fruits = []

# Add items
fruits.append("apple")
fruits.append("banana")
fruits.append("cherry")
print(f"Fruits list: {fruits}")

# Access items
print(f"First fruit: {fruits[0]}")
print(f"Last fruit: {fruits[-1]}")

# Slicing
# Get items from index 1 up to (but not including) index 3
some_fruits = fruits[1:3]
print(f"Some fruits (index 1 and 2): {fruits[1:3]}")

# List of mixed types
mixed_list = [1, "hello", 3.14, True]
print(f"Mixed list: {mixed_list}")

# Loop through a list
print("Iterating through fruits:")
for fruit in fruits:
    print(fruit)

# Accessing an index that doesn't exist will cause an IndexError
# print(fruits[10]) # Uncomment to see IndexError
```


**More List Methods/Functions:**
*   `isinstance(variable, type)`: Checks if a variable is of a certain type (e.g., `isinstance(x, str)`).
*   `enumerate(my_list)`: Used in loops to get both the index and the item.
*   `my_list.sort()`: Sorts the list in place (modifies the original list). Works if items are comparable (e.g., all numbers or all strings).
*   `len(my_list)`: Gets the number of items in the list.
*   `my_list.count(item)`: Counts how many times an item appears in the list.

**Example (More Complex):**



```python
data = ["John", 3234, "Eric", 234, "Jessica", "Lois", 2384, "Eric"]
print(f"Original data: {data}")
print(f"Number of items in original data: {len(data)}")

# Let's extract only the names (strings)
names_only = []
for item in data:
    if isinstance(item, str): # Check if the item is a string
        names_only.append(item)

print(f"Names only (before sorting): {names_only}")

# Sort the names
names_only.sort()
print(f"Names only (sorted): {names_only}")
print(f"Number of names: {len(names_only)}")

# Print names with a number
print("Numbered sorted names:")
for index, name in enumerate(names_only):
    print(f"{index + 1}. {name}") # index + 1 because enumerate is 0-based

# Count occurrences of "Eric"
eric_count = names_only.count("Eric")
print(f"'Eric' appears {eric_count} times in the sorted names list.")
```


**4. Dictionaries**
Dictionaries are unordered (in Python versions before 3.7, ordered in 3.7+) collections of `key:value` pairs.
They are mutable and very useful for storing structured data.
Keys must be unique and are typically strings or numbers. Values can be any data type (including other dictionaries or lists).
Dictionaries are defined using curly braces `{}`.

**Key Dictionary Operations:**
*   Creating an empty dictionary: `my_dict = {}` or `my_dict = dict()`
*   Adding/Accessing items: `my_dict['key'] = value`, `value = my_dict['key']`
*   Iterating: `for key, value in my_dict.items():`
*   Deleting items: `del my_dict['key']` or `value = my_dict.pop('key')` (`.pop()` returns the value)
*   Safe access with `.get()`: `value = my_dict.get('key', default_value_if_not_found)`

**Example:**



```python
# Create an empty dictionary
phonebook = {}

# Add entries
phonebook["John"] = {"Phone": "012-794-7940", "Email": "john@example.com"}
phonebook["Jill"] = {"Phone": "012-345-3450", "Email": "jill@example.com"}
phonebook["Joss"] = {"Phone": "012-321-3210", "Email": "joss@example.com"}

print(f"Full phonebook: {phonebook}")
print(f"John's email: {phonebook['John']['Email']}")

# Iterate through the dictionary
print("\nPhonebook entries:")
for name, record in phonebook.items():
    print(f"{name}'s phone is {record['Phone']}, and email is {record['Email']}")

# Delete an entry using 'del'
del phonebook["John"]
print(f"\nPhonebook after deleting John: {phonebook}")

# Delete an entry using 'pop' (and get the value)
jill_record = phonebook.pop("Jill")
print(f"Popped Jill's record: {jill_record}")
print(f"Phonebook after popping Jill: {phonebook}")

# Trying to access a non-existent key directly will cause a KeyError
# print(phonebook['NonExistent']) # Uncomment to see KeyError

# Safe access using .get()
non_existent_record = phonebook.get("NonExistent") # Returns None if not found
print(f"\nRecord for 'NonExistent' (using .get()): {non_existent_record}")

joss_record_safe = phonebook.get("Joss", "Record not found")
print(f"Record for 'Joss' (using .get()): {joss_record_safe}")

default_record = phonebook.get("AlsoNonExistent", "No entry here!")
print(f"Record for 'AlsoNonExistent' with default: {default_record}")
```

**Basic Operators**

Operators are symbols that perform operations on variables and values.

**1. Arithmetic Operators**
Used with numeric values to perform common mathematical operations:
*   `+` (Addition)
*   `-` (Subtraction)
*   `*` (Multiplication)
*   `/` (Division - always results in a float)

Python follows standard order of operations (PEMDAS/BODMAS).

**Logical Conditions and Control Flow**

Control flow statements allow you to execute certain blocks of code conditionally or repeatedly.

**`if`, `elif`, `else` Statements**
Used for conditional execution.
*   `if condition:`: Executes block if condition is `True`.
*   `elif another_condition:` (else if): Executes block if previous `if`/`elif` were `False` and `another_condition` is `True`.
*   `else:`: Executes block if all preceding `if`/`elif` conditions were `False`.

**Logical Operators:**
*   `and`: `True` if both operands are true.
*   `or`: `True` if at least one operand is true.
*   `not`: Inverts a boolean value (`not True` is `False`).
*   `in`: Checks if an item is present in a sequence (list, tuple, string).
*   `is`: Checks if two variables refer to the same object in memory (stricter than `==` which checks for equal value).

**Example (if/elif/else):**
Pay attention to indentation!



```python
x = 2
print(x == 2)
print(x == 3)
print(x < 3)

name = "John"
age = 25
if name == "John" and age > 18:
    print("John is an adult.")

if name == "John" or name == "Jill":
    print("Name is John or Jill.")

valid_names = ["John", "Jill", "Jess"]
if name in valid_names:
    print(f"{name} is a valid name.")

temperature = 22
if temperature > 30:
    print("It's hot!")
elif temperature > 20:
    print("It's warm.")
elif temperature > 10:
    print("It's cool.")
else:
    print("It's cold!")

if x == 2:
    print("x == 2 (this is correctly formatted and part of the if block)")

x_val = 2
y_val = 10
if x_val > 2:
    print("x_val > 2")
elif x_val == 2 and y_val > 50:
    print("x_val == 2 and y_val > 50")
elif x_val < 10 or y_val > 50:
    print("x_val < 10 or y_val > 50")
else:
    print("Nothing worked.")

list1 = ["John", "Jill"]
list2 = ["John", "Jill"]
list3 = list1

print(f"not (list1 == list2): {not (list1 == list2)}")
print(f"list1 == list2: {list1 == list2}")
print(f"list1 is list2: {list1 is list2}")
print(f"list1 is list3: {list1 is list3}")
```

**Loops**

Loops are used to execute a block of code repeatedly.

**`for` Loops**
Iterates over a sequence (like a list, string, or range).
*   `range(stop)`: Generates numbers from 0 up to (but not including) `stop`.
*   `range(start, stop)`: Generates numbers from `start` up to `stop`.
*   `range(start, stop, step)`: Generates numbers from `start` up to `stop`, incrementing by `step`.
*   `enumerate(sequence)`: Returns an iterator of (index, value) pairs.

**Example:**



```python
print("For loop with enumerate and range(2, 8, 2):")
for i, x_val in enumerate(range(2, 8, 2)):
    print(f"Iteration {i+1}. Range value: {x_val}")

print("\nIterating through a list of strings:")
colors = ["red", "green", "blue"]
for color in colors:
    print(color)
```


**`while` Loops**
Executes a block of code as long as a condition is `True`.
Can have an optional `else` block that executes when the loop condition becomes `False` (and the loop wasn't terminated by a `break`).
A shorthand for `count = count + 1` is `count += 1`.

**Example:**



```python
print("While loop example:")
count = 0
while count < 5:
    print(f"Count is: {count}")
    count += 1
else:
    print("End of while loop reached (count is no longer < 5).")
```


**`break` and `continue`**
*   `break`: Exits the current loop immediately.
*   `continue`: Skips the rest of the current iteration and proceeds to the next iteration of the loop.

**Example:**



```python
print("Break and while conditional example:")
count = 0
while True:
    print(f"Current count (break example): {count}")
    count += 1
    if count >= 5:
        print("Breaking out of the loop.")
        break

print("\nContinue example (print even numbers up to 7):")
for x_val in range(8):
    if x_val % 2 != 0:
        continue
    print(f"Even number: {x_val}")
```

**List Comprehensions**

A concise way to create lists. Often used to apply an operation to each item in a sequence and/or filter items.
Basic syntax: `new_list = [expression for item in iterable if condition]`
(The `if condition` part is optional).

**Example:**
Imagine we have a sentence and want to count the length of each word, skipping "the".

**Traditional way:**



```python
sentence = "for the song and the sword are birthrights sold to an usurer, but I am the last lone highwayman and I am the last adventurer"
words = sentence.split()
word_lengths_traditional = []
for word in words:
    if word != "the":
        word_lengths_traditional.append(len(word))
print(f"Word lengths (traditional): {word_lengths_traditional}")

# Using List Comprehension:
word_lengths_comprehension = [len(word) for word in words if word != "the"]
print(f"Word lengths (comprehension): {word_lengths_comprehension}")

squares = [x**2 for x in range(10)]
print(f"Squares: {squares}")

even_squares = [x**2 for x in range(10) if x % 2 == 0]
print(f"Even squares: {even_squares}")
```

**Exception Handling (`try`, `except`, `raise`)**

Exceptions are errors that occur during program execution.
Handling exceptions allows your program to manage errors gracefully instead of crashing.
*   `try`: The block of code to attempt.
*   `except ExceptionType:`: If an `ExceptionType` occurs in the `try` block, this block is executed. You can catch specific exceptions (e.g., `ValueError`, `TypeError`, `IndexError`) or general `Exception`. It's good practice to be specific.
*   `else`: (Optional) Executes if the `try` block completes without raising an exception.
*   `finally`: (Optional) Always executes, whether an exception occurred or not. Useful for cleanup.
*   `raise ExceptionType("message")`: Manually triggers an exception.

**Example: Catching an error**
Trying to add an integer and a string:



```python
try:
    result = 1 + "hello"
    print(f"Result: {result}")
except TypeError:
    print("Oops! A TypeError occurred. Cannot add an integer and a string directly.")
except ValueError:
    print("A ValueError occurred.")
print("Program continues after try-except block.")

def print_list_item(my_list, index_to_access):
    try:
        print(my_list[index_to_access])
    except IndexError:
        print(f"Index {index_to_access} is out of bounds for list {my_list}. Printing 0 instead.")
        print(0)

sample_list = [1, 2, 3, 4, 5, 6, 7]
print_list_item(sample_list, 3)
print_list_item(sample_list, 10)

def check_positive(number):
    if number <= 0:
        raise ValueError("Number must be positive!")
    print(f"{number} is a positive number.")

try:
    check_positive(10)
    check_positive(-5)
    check_positive(20)
except ValueError as e:
    print(f"Caught a ValueError: {e}")
```

**Sets**

Sets are unordered collections of unique items. They are mutable.
Duplicates are automatically removed when creating a set or adding items.
Useful for membership testing, removing duplicates from a sequence, and set mathematics (union, intersection, etc.).
Defined using curly braces `{}` (like dictionaries, but without key:value pairs) or `set()`.
An empty set MUST be created with `set()`, because `{}` creates an empty dictionary.

**Key Set Operations:**
*   Creating a set: `my_set = {1, 2, 3}` or `my_set = set([1, 2, 2, 3])` (results in `{1, 2, 3}`)
*   `set1.intersection(set2)`: Items common to both sets.
*   `set1.union(set2)`: All items from both sets (duplicates removed).
*   `set1.difference(set2)`: Items in `set1` but not in `set2`.
*   `set1.symmetric_difference(set2)`: Items in either `set1` or `set2`, but not in both.

**Example:**



```python
sentence = "the rain is wet and wet is the rain"
words_in_sentence = sentence.split()
unique_words = set(words_in_sentence)
print(f"Original words list: {words_in_sentence}")
print(f"Unique words set: {unique_words}")

set_one = {"Alice", "Carol", "Dan", "Eve", "Heidi"}
set_two = {"Bob", "Dan", "Eve", "Grace", "Heidi"}

print(f"\nSet One: {set_one}")
print(f"Set Two: {set_two}")

intersection_result = set_one.intersection(set_two)
print(f"Intersection (common to both): {intersection_result}")

union_result = set_one.union(set_two)
print(f"Union (all unique items from both): {union_result}")

difference_one_two = set_one.difference(set_two)
print(f"Difference (Set One - Set Two): {difference_one_two}")

difference_two_one = set_two.difference(set_one)
print(f"Difference (Set Two - Set One): {difference_two_one}")

symmetric_diff = set_one.symmetric_difference(set_two)
print(f"Symmetric Difference: {symmetric_diff}")
```
